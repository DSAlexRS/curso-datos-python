## 2. Condicionales

`if` evalúa una condición. El bloque indentado se ejecuta únicamente cuando es verdadera. La indentación de cuatro espacios forma parte de la sintaxis.


```python
porcentaje_internet = 73.0
if porcentaje_internet >= 70:
    print("Cumple el umbral")
print("Esta línea está fuera del if")
```

`else` define qué ocurre cuando la condición es falsa. `elif` permite evaluar alternativas ordenadas. Python se detiene en la primera verdadera.


```python
porcentaje_internet = 63.0
if porcentaje_internet >= 85:
    categoria = "Alta"
elif porcentaje_internet >= 70:
    categoria = "Media"
else:
    categoria = "Baja"
print(categoria)
```

El orden importa. Si comenzáramos por `>= 70`, un valor de 90 entraría allí y nunca llegaría a `>= 85`.

## 3. Ciclo `for`

`for` toma cada elemento de una colección. La variable temporal recibe un valor diferente en cada vuelta.


```python
nombres = ["Colombia", "México", "Perú", "Chile"]
# Recorremos la lista; en cada vuelta, nombre contiene un país distinto.
for nombre in nombres:
    print("País analizado:", nombre)
```

En una lista de diccionarios, cada `registro` es un diccionario. Podemos consultar sus claves dentro del ciclo.


```python
# Cada registro es un diccionario con los datos de un país.
for registro in paises:
    print(registro["pais"], "->", registro["internet"])
```

### Acumuladores y contadores

Un acumulador agrega valores. Un contador aumenta cuando se cumple una regla. Ambos se definen antes del ciclo.


```python
# 1. Inicializamos las variables antes de recorrer los registros.
suma_internet = 0
cantidad = 0
# 2. Acumulamos el valor y contamos una observación por vuelta.
for registro in paises:
    suma_internet += registro["internet"]
    cantidad += 1
# 3. Calculamos el promedio cuando el ciclo ya terminó.
promedio_internet = suma_internet / cantidad
print("Suma:", suma_internet)
print("Cantidad:", cantidad)
print("Promedio:", round(promedio_internet, 2))
```

```python
# El contador aumenta únicamente cuando se cumple la condición.
cantidad_alta = 0
for registro in paises:
    if registro["internet"] >= 80:
        cantidad_alta += 1
print("Países con 80% o más:", cantidad_alta)
```

### Pausa práctica: promedio ignorando faltantes

**Tiempo sugerido: 3–5 minutos.** Suma únicamente las tasas de desempleo disponibles y calcula el promedio usando un contador válido.


```python
suma_desempleo = 0
cantidad_valida = 0
# Completa el ciclo
```

### Respuesta sugerida


```python
suma_desempleo = 0
cantidad_valida = 0
# Sumamos y contamos únicamente observaciones disponibles.
for registro in paises:
    if registro["desempleo"] is not None:
        suma_desempleo += registro["desempleo"]
        cantidad_valida += 1
# Calculamos después de terminar el recorrido.
promedio_desempleo = suma_desempleo / cantidad_valida
print(round(promedio_desempleo, 2))
```

### Construir una nueva lista

Podemos comenzar con una lista vacía y agregar solo las observaciones que cumplan una regla. Este patrón anticipa los filtros de Pandas.


```python
# Creamos una lista nueva sin modificar la lista original.
paises_seleccionados = []
for registro in paises:
    if registro["internet"] >= 75:
        paises_seleccionados.append(registro["pais"])
print(paises_seleccionados)
```

## 4. `range()`, `enumerate()` y `zip()`

## 5. Ciclo `while`

`while` repite mientras una condición sea verdadera. Es útil cuando no conocemos el número de repeticiones, por ejemplo al consultar páginas de una API. Siempre debe existir una instrucción que acerque la condición a `False`.


```python
# pagina cambia en cada vuelta para evitar un ciclo infinito.
pagina = 1
while pagina <= 3:
    print("Consultando página", pagina)
    pagina += 1
print("Consulta terminada")
```

`break` interrumpe el ciclo. Conviene usarlo con un criterio de salida claro y documentado.


```python
# El ciclo puede terminar por el límite o por encontrar una respuesta.
intentos = 1
while intentos <= 5:
    print("Intento", intentos)
    if intentos == 3:
        print("Se obtuvo respuesta")
        break
    intentos += 1
```

```python
numero = 2
# Completa el while
```

### Respuesta sugerida


```python
numero = 2
# Aumentamos numero en cada vuelta para terminar el ciclo.
while numero <= 8:
    print(numero)
    numero += 2
```

## 6. Funciones

Una función reúne instrucciones relacionadas. Se define con `def`, recibe datos mediante parámetros y puede devolver un resultado con `return`. Definirla no la ejecuta: hay que llamarla.


```python
def mostrar_bienvenida():
    # Cuerpo de la función: se ejecuta solo cuando la llamamos.
    print("Bienvenidos al análisis de datos")

mostrar_bienvenida()
```

### Parámetros y argumentos

Un **parámetro** es el nombre usado en la definición. Un **argumento** es el valor concreto entregado al llamar la función.


```python
def presentar_indicador(nombre, unidad):
    # Construimos el texto con los dos argumentos recibidos.
    mensaje = f"Indicador: {nombre}. Unidad: {unidad}."
    return mensaje

resultado = presentar_indicador("Desempleo", "% de la fuerza laboral")
print(resultado)
```

`print()` muestra; `return` entrega un valor para guardarlo o usarlo en otro cálculo.


```python
def variacion_porcentual(valor_inicial, valor_final):
    # Calculamos el cambio relativo respecto al valor inicial.
    cambio = (valor_final - valor_inicial) / valor_inicial * 100
    return cambio

variacion = variacion_porcentual(100, 112)
print(round(variacion, 1), "%")
```

Un valor predeterminado hace opcional un argumento. Los parámetros obligatorios aparecen primero.


```python
def convertir_a_porcentaje(proporcion, decimales=1):
    # decimales tiene un valor predeterminado, pero puede cambiarse.
    porcentaje = proporcion * 100
    return round(porcentaje, decimales)

print(convertir_a_porcentaje(0.481))
print(convertir_a_porcentaje(0.481, 2))
```

## 7. Errores frecuentes y validación

`NameError` indica un nombre inexistente; `TypeError` una operación incompatible; `KeyError` una clave ausente; `IndexError` una posición inexistente; `SyntaxError` una instrucción incompleta.

En clase podemos leer errores, pero aquí evitamos celdas que detengan la ejecución completa.


```python
ejemplo_error = "int('diez') produciría ValueError"
print(ejemplo_error)
print("Se deja como texto para que el notebook continúe.")
```

`try` intenta una operación y `except` define qué hacer ante un error esperado. No debe usarse para ocultar cualquier problema.


```python
valor_texto = "10.5"
try:
    valor_numero = float(valor_texto)
    print("Conversión correcta:", valor_numero)
except ValueError:
    print("El texto no representa un número")
```

```python
def variacion_segura(valor_inicial, valor_final):
    # Validamos ausencias y división por cero antes de calcular.
    if valor_inicial is None or valor_final is None:
        return None
    if valor_inicial == 0:
        return None
    return (valor_final - valor_inicial) / valor_inicial * 100

print(variacion_segura(100, 112))
print(variacion_segura(0, 112))
print(variacion_segura(None, 112))
```

```python
def promedio_seguro(valores):
    # Validamos la lista antes de dividir por su cantidad de elementos.
    if len(valores) == 0:
        return None
    return sum(valores) / len(valores)

print(promedio_seguro([]))
print(promedio_seguro([2, 4, 6]))
```

## 9. NumPy

Una lista y un arreglo de NumPy no se comportan igual. `lista1 + lista2` concatena; `arreglo1 + arreglo2` suma elemento a elemento.


```python
import numpy as np

lista_1 = [1, 2, 3]
lista_2 = [10, 20, 30]
print("Listas:", lista_1 + lista_2)

arreglo_1 = np.array(lista_1)
arreglo_2 = np.array(lista_2)
print("Arreglos:", arreglo_1 + arreglo_2)
```

```python
tasas = np.array([10.2, 2.8, 5.4, 8.7])
print("Tipo:", type(tasas))
print("Forma:", tasas.shape)
print("Media:", tasas.mean())
print("Mediana:", np.median(tasas))
print("Desviación estándar poblacional:", tasas.std())
```

Los arreglos permiten operaciones vectorizadas: una expresión actúa sobre todos los elementos.


```python
proporciones = np.array([0.73, 0.786, 0.747, 0.902])
porcentajes = proporciones * 100
mayores_75 = porcentajes >= 75
print(porcentajes)
print(mayores_75)
print(porcentajes[mayores_75])
```

```python
matriz = np.array([
    [73.0, 10.2],
    [78.6, 2.8],
    [74.7, 5.4],
    [90.2, 8.7],
])
print("Forma:", matriz.shape)
print("Primera fila:", matriz[0])
print("Columna Internet:", matriz[:, 0])
print("Promedio por columna:", matriz.mean(axis=0))
```

## 10. Síntesis: una regla reproducible

El siguiente bloque combina observaciones, ciclo, validación y función. Busca dejar visible la regla, no escribir el código más corto.


```python
def clasificar_desempleo(valor):
    # Atendemos primero la ausencia y luego evaluamos de mayor a menor.
    if valor is None:
        return "Sin dato"
    elif valor >= 10:
        return "Alta"
    elif valor >= 6:
        return "Media"
    return "Baja"

# Aplicamos la función a cada registro y guardamos un resultado resumido.
resultados = []
for registro in paises:
    resultado = {
        "pais": registro["pais"],
        "nivel_desempleo": clasificar_desempleo(registro["desempleo"]),
    }
    resultados.append(resultado)
print(resultados)
```
